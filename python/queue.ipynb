```python
class queue:
    def __init__ (self):
        self.queue=[]
        
    def enqueue(self,item):
        self.queue.append(item)
        
    def dequeue(self):
        if len(self.queue)==0:
            print("queue is empty")
        else:
            return self.queue.pop(0)
    def top(self):
        if len(self.queue)==0:
            print('queue deosent exist')
        else:
            return self.queue[0]
    def display(self):
        print("queue elements",self.queue)

    def is_empty(self):
        return len(self.queue)==0
```

```python
q1=queue()
```

```python
q1.dequeue()
```

```python
q1.enqueue(20)
q1.enqueue(30)
q1.enqueue(60)
```

```python
q1.display()
```
