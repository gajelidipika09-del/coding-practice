```python
class stack:
    def __init__ (self):
        self.stack=[]
        
    def push(self,item):
        self.stack.append(item)
        
    def pop(self):
        if len(self.stack)==0:
            print("stack is empty")
        else:
            return self.stack.pop()
    def top(self):
        if len(self.stack)==0:
            print('stack deosent exist')
        else:
            return self.stack[-1]
    def display(self):
        print("stack elements",self.stack)

    def is_empty(self):
        return len(self.stack)==0
```
